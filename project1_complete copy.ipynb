## Task 1.1 – Discretización del Mundo

````python
# La imagen tiene miles de píxeles. En lugar de tratar cada píxel como un nodo (ineficiente), la dividimos en "tiles" de `tile_size x tile_size` píxeles.
# Cada tile se convierte en 
# un nodo del grafo.

FREE  = 0   # blanco (libre)
WALL  = 1   # negro (pared)
START = 2   # rojo (inicio)
GOAL  = 3   # verde (metas)

def classify_tile(tile_pixels: np.ndarray) -> int:
    # Promedio de R, G, B en todos los píxeles del tile
    avg = tile_pixels.mean(axis=(0, 1))  # shape: (3,)
    r, g, b = avg[0], avg[1], avg[2]
    
    # Clasificacion de color
    if r < 50 and g < 50 and b < 50:
        return WALL
    elif r > 150 and g < 80 and b < 80:
        return START
    elif g > 150 and r < 80 and b < 80:
        return GOAL
    else:
        return FREE


def discretize_image(image_path: str, tile_size: int = 2):
    # 1. Cargar imagen y convertir a RGB
    img = Image.open(image_path).convert('RGB')
    img_array = np.array(img)  # shape: (alto, ancho, 3)
    H, W, _ = img_array.shape
    print(f"Imagen cargada: {W}x{H} píxeles")
    
    # 2. Calcular dimensiones de la grilla
    rows = H // tile_size
    cols = W // tile_size
    print(f"Grilla discreta: {rows}x{cols} nodos (tile_size={tile_size})")
    
    # 3. Construir la matriz
    grid = np.zeros((rows, cols), dtype=int)
    start = None
    goals = []
    
    for r in range(rows):
        for c in range(cols):
            # Extraer los píxeles de este tile
            r0, r1 = r * tile_size, (r + 1) * tile_size
            c0, c1 = c * tile_size, (c + 1) * tile_size
            tile = img_array[r0:r1, c0:c1]  # shape: (tile_size, tile_size, 3)
            
            # Clasificar el tile
            cell_type = classify_tile(tile)
            grid[r, c] = cell_type
            
            # Registrar inicio y metas
            if cell_type == START:
                start = (r, c)
            elif cell_type == GOAL:
                goals.append((r, c))
    
    print(f"Inicio encontrado en: {start}")
    print(f"Metas encontradas en: {goals}")
    
    return grid, start, goals, img_array


def visualize_grid(grid: np.ndarray, title: str = "Grilla Discreta"):
    # Mapa de colores: 0=blanco, 1=negro, 2=rojo, 3=verde
    color_map = {
        FREE:  [1.0, 1.0, 1.0],   # Blanco
        WALL:  [0.0, 0.0, 0.0],   # Negro
        START: [1.0, 0.0, 0.0],   # Rojo
        GOAL:  [0.0, 0.8, 0.0],   # Verde
    }
    
    rows, cols = grid.shape
    img_vis = np.ones((rows, cols, 3))
    for cell_type, color in color_map.items():
        mask = grid == cell_type
        img_vis[mask] = color
    
    plt.figure(figsize=(8, 8))
    plt.imshow(img_vis, interpolation='nearest')
    plt.title(title)
    plt.axis('off')
    
    # Leyenda
    patches = [
        mpatches.Patch(color='white', label='Libre', edgecolor='gray'),
        mpatches.Patch(color='black', label='Pared'),
        mpatches.Patch(color='red',   label='Inicio'),
        mpatches.Patch(color='green', label='Meta'),
    ]
    plt.legend(handles=patches, loc='upper right')
    plt.tight_layout()
    plt.show()
````

## Task 1.2 – Framework OOP y Búsqueda (BFS / DFS)

### Framework OOP

````python
# clase base Problem que define qué debe poder hacer cualquier problema de búsqueda
# MazeProblem implementa ese contrato para nuestro laberinto.

class Problem(ABC): # Clase abstracta que representa cualquier problema de busqueda
    
    def __init__(self, initial_state): # Donde empieza el agente
        self.initial_state = initial_state
    
    @abstractmethod
    def actions(self, state): #Returns list[str] de acciones válidas desde 'state'
        pass
    
    @abstractmethod
    def result(self, state, action): # Returns el nuevo estado tras aplicar 'action' to 'state'
        pass
    
    @abstractmethod
    def goal_test(self, state) -> bool: # Returns True si 'state' es un estado meta
        pass
    
    def step_cost(self, state, action, next_state) -> float: # Costo de moverse
        return 1.0 # default 1


class MazeProblem(Problem): # Implementa Problem para un laberinto en grilla 2D, el estado es una tupla (fila, columna)
    
    # Las 4 acciones posibles: (delta_fila, delta_col)
    DIRECTIONS = {
        'UP':    (-1,  0),
        'DOWN':  ( 1,  0),
        'LEFT':  ( 0, -1),
        'RIGHT': ( 0,  1),
    }
    
    def __init__(self, grid: np.ndarray, start: tuple, goals: list):
        super().__init__(initial_state=start)
        self.grid  = grid
        self.goals = set(goals)  # Set para búsqueda O(1)
        self.rows, self.cols = grid.shape
    
    def actions(self, state: tuple) -> list:
        r, c = state
        valid = []
        for action, (dr, dc) in self.DIRECTIONS.items():
            nr, nc = r + dr, c + dc
            # Verificar límites del mapa sino no cuenta la acción
            if 0 <= nr < self.rows and 0 <= nc < self.cols:
                # Verificar que no sea pared
                if self.grid[nr, nc] != WALL:
                    valid.append(action)
        return valid
    
    def result(self, state: tuple, action: str) -> tuple: # Aplica la acción y retorna el nuevo estado (posición).
        r, c = state
        dr, dc = self.DIRECTIONS[action]
        return (r + dr, c + dc)
    
    def goal_test(self, state: tuple) -> bool: # Returns True si el estado es una de las metas.
        return state in self.goals
    
    def step_cost(self, state, action, next_state) -> float:
        return 1.0
````

### Búsqueda

````python
class Node:
    
    def __init__(self, state, parent=None, action=None, g_cost=0.0):
        self.state   = state # estado actual (posición en el laberinto) (x, y)
        self.parent  = parent # nodo padre (para reconstruir el camino)
        self.action  = action # acción que llevó a este nodo
        self.g_cost  = g_cost # costo acumulado desde el inicio hasta aquí
    
    def path(self) -> list: # reconstruye el camino desde el nodo inicial hasta este nodo
        states = []
        node = self
        while node is not None:
            states.append(node.state)
            node = node.parent
        states.reverse()  # De inicio a fin
        return states
    
    # Necesario para que el heap de A* pueda comparar nodos
    def __lt__(self, other):
        return self.g_cost < other.g_cost
````

````python
# La lógica central que todos los algoritmos comparten.
# La diferencia entre BFS, DFS y A* está en la frontera.

def graph_search(problem: Problem, frontier): # problema y frontera

    # Nodo inicial
    start_node = Node(state=problem.initial_state)
    frontier.push(start_node)
    
    # Conjunto de estados ya visitados (evita ciclos)
    explored = set()
    
    while not frontier.is_empty():
        node = frontier.pop()
        
        # Verificar si llegamos a la meta
        if problem.goal_test(node.state):
            return node.path()
        
        # Si ya visitamos este estado, lo saltamos
        if node.state in explored:
            continue
        explored.add(node.state)
        
        # Expandir: generar hijos
        for action in problem.actions(node.state):
            next_state = problem.result(node.state, action)
            if next_state not in explored:
                cost = node.g_cost + problem.step_cost(node.state, action, next_state)
                child = Node(state=next_state, parent=node, action=action, g_cost=cost)
                frontier.push(child)
    
    return None  # No se encontró solución
    # Return lista de estados (el camino)


# Fronteras
class FIFOQueue: # implementa BFS, garantiza explorar por niveles (Camino mas corto)

    def __init__(self):
        self.queue = deque()
    
    def push(self, node):
        self.queue.append(node)   # Agrega al final
    
    def pop(self):
        return self.queue.popleft()  # Saca del frente
    
    def is_empty(self):
        return len(self.queue) == 0


class LIFOStack: # implementa DFS, garantiza explorar un camino profundo antes de retroceder

    def __init__(self):
        self.stack = []
    
    def push(self, node):
        self.stack.append(node)   # Agrega al tope
    
    def pop(self):
        return self.stack.pop()   # Saca del tope
    
    def is_empty(self):
        return len(self.stack) == 0

#Busquedas
def bfs(problem: Problem): #camino mas corto
    print("Ejecutando BFS...")
    path = graph_search(problem, FIFOQueue())
    if path:
        print(f"BFS encontró camino con {len(path)-1} pasos.")
    else:
        print("BFS: No se encontró solución.")
    return path


def dfs(problem: Problem): #explora rápido, no garantiza camino óptimo
    print("Ejecutando DFS...")
    path = graph_search(problem, LIFOStack())
    if path:
        print(f"DFS encontró camino con {len(path)-1} pasos.")
    else:
        print("DFS: No se encontró solución.")
    return path
````

## Task 1.3 – A* (A-Star)
```
f(n) = g(n) + h(n)
```
- `g(n)` = costo real acumulado desde el inicio hasta el nodo `n`
- `h(n)` = estimación heurística (Distancia Manhattan al goal más cercano)
- `f(n)` = costo total estimado del camino pasando por `n`

````python
# Frontera para A*: Priority Queue (Min-Heap)
# A* usa información heurística para guiar la búsqueda hacia la meta de forma inteligente.

class PriorityQueue: # Clase para la frontera, implementada como una cola de prioridad minima usando heapq.

    def __init__(self):
        self.heap = []
        self.counter = 0  # Desempate cuando dos nodos tienen igual prioridad
    
    def push(self, node, priority: float): # Agrega un nodo a la cola de prioridad
        heapq.heappush(self.heap, (priority, self.counter, node))
        self.counter += 1
    
    def pop(self): # Extrae el nodo con menor prioridad
        _, _, node = heapq.heappop(self.heap)
        return node
    
    def is_empty(self):
        return len(self.heap) == 0


def heuristic_manhattan(state: tuple, goals: list) -> float: # Heurística de distancia Manhattan al goal más cercano
    r, c = state
    return min(abs(r - gr) + abs(c - gc) for gr, gc in goals)


def astar(problem: MazeProblem) -> list: # Algoritmo A* para el laberinto

    print("Ejecutando A*...")
    
    goals_list = list(problem.goals)
    frontier = PriorityQueue()
    
    # Nodo inicial con f = g + h = 0 + h(start)
    start_node = Node(state=problem.initial_state, g_cost=0.0)
    h0 = heuristic_manhattan(problem.initial_state, goals_list)
    frontier.push(start_node, priority=h0)  # f = 0 + h
    
    explored = set()
    
    while not frontier.is_empty():
        node = frontier.pop()
        
        if problem.goal_test(node.state):
            print(f"A* encontró camino con {len(node.path())-1} pasos, costo total = {node.g_cost:.1f}")
            return node.path()
        
        if node.state in explored:
            continue
        explored.add(node.state)
        
        for action in problem.actions(node.state):
            next_state = problem.result(node.state, action)
            if next_state not in explored:
                g_new = node.g_cost + problem.step_cost(node.state, action, next_state)
                h_new = heuristic_manhattan(next_state, goals_list)
                f_new = g_new + h_new
                child = Node(state=next_state, parent=node, action=action, g_cost=g_new)
                frontier.push(child, priority=f_new)
    
    print("A*: No se encontró solución.")
    return None
````

## Task 2.1 – Entrenamiento de la Red Neuronal

````python
import pandas as pd

# ============================================================
# MAPEO DE COLORES A COSTOS
# Justificación: refleja riesgo y dificultad de travesía para un robot de entrega.
# Grey/White = pavimento (costo 1), Green = grama (3), Brown = tierra (4),
# Yellow = arena (5), Orange = arena gruesa (6), Pink/Purple = terreno indefinido (7),
# Red = nodo de inicio, costo neutro (1), Blue = agua, alto riesgo (10),
# Black = pared/obstáculo, nunca transitable (99).
# ============================================================
COLOR_COST_MAP = {
    'Grey':   1,
    'White':  1,
    'Green':  3,
    'Brown':  4,
    'Yellow': 5,
    'Orange': 6,
    'Pink':   7,
    'Purple': 7,
    'Red':    1,   # Nodo de inicio → costo neutro
    'Blue':   10,
    'Black':  99,  # Seguridad: nunca debe ocurrir (filtrado antes por is_wall)
}
````

````python
# ============================================================
# UTILIDADES: encoding y split manual
# ============================================================

def encode_labels(y_raw):
    """Convierte etiquetas de texto a índices y one-hot encoding."""
    unique_labels = sorted(set(y_raw))
    label_to_index = {label: i for i, label in enumerate(unique_labels)}
    index_to_label = {i: label for label, i in label_to_index.items()}

    n = len(y_raw)
    num_classes = len(unique_labels)
    y_idx = np.array([label_to_index[label] for label in y_raw])
    y_onehot = np.zeros((n, num_classes))
    y_onehot[np.arange(n), y_idx] = 1.0

    return y_onehot, y_idx, label_to_index, index_to_label


def train_test_split_manual(X, y_oh, y_idx, test_ratio=0.2, seed=42):
    """Split manual estratificado (sin sklearn)."""
    rng = np.random.default_rng(seed)
    n = len(X)
    indices = rng.permutation(n)
    split = int(n * (1 - test_ratio))
    train_idx, test_idx = indices[:split], indices[split:]
    return (X[train_idx], X[test_idx],
            y_oh[train_idx], y_oh[test_idx],
            y_idx[train_idx], y_idx[test_idx])
````

````python
# ============================================================
# CARGAR DATASET REAL
# Dataset: Associated RGBs and Basic Color Names
# Columnas: red, green, blue, label  (o R, G, B, label)
# ============================================================

df = pd.read_csv('final_data_colors.csv')  # <-- Ajusta la ruta si es necesario

# Normalizar nombres de columnas (acepta tanto minúsculas como mayúsculas)
df.columns = [c.strip() for c in df.columns]
col_map = {c.lower(): c for c in df.columns}
df = df.rename(columns={
    col_map.get('red',   'red'):   'R',
    col_map.get('green', 'green'): 'G',
    col_map.get('blue',  'blue'):  'B',
})

print(f"Dataset: {len(df)} muestras")
print(df['label'].value_counts())

X_all  = df[['R', 'G', 'B']].values.astype(np.float64) / 255.0
y_raw  = df['label'].values

y_onehot, y_idx, label_to_index, index_to_label = encode_labels(y_raw)
X_train, X_test, y_oh_train, y_oh_test, y_idx_train, y_idx_test = train_test_split_manual(
    X_all, y_onehot, y_idx
)

print(f"\nTrain: {len(X_train)} | Test: {len(X_test)}")
print(f"Clases ({len(label_to_index)}): {list(label_to_index.keys())}")
````

````python
# ============================================================
# MLP – Implementación desde cero con numpy
# Arquitectura: 3 → 64 → 32 → num_classes
# Activación: ReLU en ocultas, Softmax en salida
# Entrenamiento: SGD + Backpropagation
# ============================================================

# multilayer perceptron
class MLP:
    """Multi-Layer Perceptron implementado únicamente con numpy."""

    def __init__(self, layer_sizes: list, lr: float = 0.01, seed: int = 42):
        """
        layer_sizes: lista con el tamaño de cada capa, incluyendo entrada y salida.
                     Ej: [3, 64, 32, 11]
        """
        rng = np.random.default_rng(seed)
        self.lr = lr
        self.weights = []
        self.biases  = []

        for i in range(len(layer_sizes) - 1):
            fan_in = layer_sizes[i]
            fan_out = layer_sizes[i + 1]
            # He initialization para ReLU
            W = rng.standard_normal((fan_in, fan_out)) * np.sqrt(2.0 / fan_in)
            b = np.zeros((1, fan_out))
            self.weights.append(W)
            self.biases.append(b)

    # --- Activaciones ---

    def relu(self, z):        return np.maximum(0, z)
    def relu_grad(self, z):   return (z > 0).astype(float)

    # Usan la función de activación ReLU, lo que permite que la red aprenda relaciones no lineales
    # entre los valores RGB y la clase del terreno.

    def softmax(self, z):
        z_shifted = z - z.max(axis=1, keepdims=True)  # Estabilidad numérica
        exp_z = np.exp(z_shifted)
        return exp_z / exp_z.sum(axis=1, keepdims=True)

    # --- Forward pass ---

    def forward(self, X):
        """Devuelve activaciones de todas las capas y la salida final."""
        self.zs = []     # Pre-activaciones
        self.acts = [X]  # Activaciones (la primera es la entrada)

        a = X
        for i, (W, b) in enumerate(zip(self.weights, self.biases)):
            z = a @ W + b
            self.zs.append(z)
            if i < len(self.weights) - 1:
                a = self.relu(z)   # Capas ocultas: ReLU
            else:
                a = self.softmax(z)  # Capa de salida: Softmax
            self.acts.append(a)

        return self.acts[-1]  # Probabilidades de clase

    # --- Loss: Cross-Entropy ---

    def cross_entropy_loss(self, y_pred, y_true_onehot):
        eps = 1e-9
        return -np.mean(np.sum(y_true_onehot * np.log(y_pred + eps), axis=1))

    # --- Backward pass (Backpropagation) ---
    # Propagación hacia atrás y Cálculo de gradientes
    def backward(self, y_true_onehot):
        """Calcula gradientes y actualiza pesos con SGD."""
        n = y_true_onehot.shape[0]
        num_layers = len(self.weights)

        # Gradiente en la salida (softmax + cross-entropy simplificado)
        delta = self.acts[-1] - y_true_onehot  # shape: (n, num_classes)

        for i in reversed(range(num_layers)):
            a_prev = self.acts[i]
            dW = (a_prev.T @ delta) / n
            db = delta.mean(axis=0, keepdims=True)

            if i > 0:  # Propagar hacia atrás
                delta = (delta @ self.weights[i].T) * self.relu_grad(self.zs[i - 1])

            # Actualización SGD
            self.weights[i] -= self.lr * dW
            self.biases[i]  -= self.lr * db

            # SGD actualiza los pesos usando los gradientes calculados:
            # 𝑊 = 𝑊 − 𝜂∇𝑊
            # donde self.lr es el learning rate.
            # Se entrena por mini-batches, lo que hace el proceso más eficiente.

    # --- Entrenamiento ---

    def fit(self, X_train, y_oh_train, epochs: int = 100, batch_size: int = 32, verbose: bool = True):
        """Entrena la red con Mini-Batch SGD."""
        n = X_train.shape[0]
        loss_history = []

        for epoch in range(epochs):
            # Mezclar datos
            perm = np.random.permutation(n)
            X_shuf, y_shuf = X_train[perm], y_oh_train[perm]

            epoch_loss = 0.0
            for i in range(0, n, batch_size):
                Xb = X_shuf[i:i + batch_size]
                yb = y_shuf[i:i + batch_size]
                y_pred = self.forward(Xb)
                epoch_loss += self.cross_entropy_loss(y_pred, yb)
                self.backward(yb)

            avg_loss = epoch_loss / (n // batch_size + 1)
            loss_history.append(avg_loss)

            if verbose and (epoch + 1) % 10 == 0:
                print(f"Epoch {epoch+1:4d}/{epochs} | Loss: {avg_loss:.4f}")

        return loss_history

    # --- Predicción ---

    def predict(self, X):
        """Devuelve el índice de clase predicho para cada muestra."""
        probs = self.forward(X)
        return np.argmax(probs, axis=1)

    def predict_label(self, rgb_normalized: np.ndarray, index_to_label: dict) -> str:
        """Predice la etiqueta de color para un RGB normalizado (shape: (1,3))."""
        idx = self.predict(rgb_normalized)[0]
        return index_to_label[idx]

    # --- Accuracy ---

    def accuracy(self, X, y_idx_true):
        preds = self.predict(X)
        return np.mean(preds == y_idx_true)
````

## Task 2.2 – A* con Inferencia de la Red Neuronal

Se crea `NNMazeProblem` que **extiende** `MazeProblem` y sobreescribe **únicamente** `step_cost`. Cuando A* evalúa un vecino, extrae el color RGB promedio del tile de la imagen y lo pasa por la red neuronal para obtener el costo real del terreno.

````python
class NNMazeProblem(MazeProblem):
    """Extiende MazeProblem: el costo de cada tile se infiere con la Red Neuronal.
    
    Hereda actions(), result(), goal_test() del Task 1 sin modificación.
    Solo sobreescribe step_cost() para usar el MLP.
    """

    def __init__(self, grid: np.ndarray, start: tuple, goals: list,
                 img_array: np.ndarray, tile_size: int,
                 mlp_model: MLP, index_to_label: dict,
                 color_cost_map: dict):
        super().__init__(grid, start, goals)
        self.img_array      = img_array
        self.tile_size      = tile_size
        self.mlp            = mlp_model
        self.index_to_label = index_to_label
        self.color_cost_map = color_cost_map

    def _tile_rgb(self, state: tuple) -> np.ndarray:
        """Extrae el color RGB promedio del tile (fila, col) de la imagen original."""
        r, c = state
        ts = self.tile_size
        tile = self.img_array[r*ts:(r+1)*ts, c*ts:(c+1)*ts]  # shape: (ts, ts, 3)
        avg_rgb = tile.mean(axis=(0, 1)) / 255.0              # Normalizado [0,1]
        return avg_rgb.reshape(1, 3)                          # shape: (1, 3)

    def is_wall(self, state: tuple) -> bool:
        """Filtra paredes negras antes de consultar la red (evita clasificar (0,0,0))."""
        r, c = state
        ts = self.tile_size
        tile = self.img_array[r*ts:(r+1)*ts, c*ts:(c+1)*ts]
        return tile.mean() < 20  # Umbral: negro puro

    def step_cost(self, state, action, next_state) -> float:
        """Infiere el tipo de terreno del tile vecino con la NN y retorna su costo."""
        # Paredes ya filtradas en actions(), pero doble seguridad
        if self.is_wall(next_state):
            return 99.0

        rgb = self._tile_rgb(next_state)
        label = self.mlp.predict_label(rgb, self.index_to_label)
        cost  = self.color_cost_map.get(label, 5.0)  # Default 5 si etiqueta desconocida
        return float(cost)
````
